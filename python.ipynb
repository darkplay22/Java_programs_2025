```python
s='abc xyz abc uyt abc'
l=s.split()
print(l)
l1=[]
for i in l:
    if i not in l1:
        l1.append(i)

print(l1)
```

```python
s='abc xyz abc uyt abc'
l=s.split()
l1={}
for i in l:
    l1[i]=l.count(i)
       
for i in l1:
    if(l1.get(i)==1):
        print(i)
```
